Numpy indexing and slicing of vectors

```python
import numpy as np
```

```python
arr=np.arange(1,11)
arr
```

```python
arr[0:5]
```

```python
arr[1:6:3]
```

```python
arr[:5]
```

```python
arr[1::]
```

Numpy Indexing and slicing of a Matrix

```python
arr=np.arange(1,31).reshape(6,5)
arr
```

```python
arr[0]
```

```python
arr[1]
```

```python
arr[0,0]
```

```python
arr[5,2]
```

```python
arr[2:4,1:3]
```

```python
arr[4:,3:]
```

```python
arr[:,2]
```

Boolean Indexing

```python
arr=np.arange(1,11)
arr
```

```python
bool_index=arr%2==0
bool_index
```

```python
arr=arr[bool_index]
arr
```
